## Dataset 1

```python
import pandas as pd

path1 = "mexico-real-estate-1.csv"
path2 = "mexico-real-estate-2 (11).csv"
path3 = "mexico-real-estate-3.csv"

MXN_PER_USD = 18
```

```python
df1 = pd.read_csv(path1)
df1.info()
df1.sample(5)
```

```python
# Drop rows with missing lat/lon
df1.dropna(inplace=True)

# Convert price from text ("$67,965.56") to float
df1["price_usd"] = (
    df1["price_usd"].astype(str)
    .str.replace(r"[$,]", "", regex=True)
    .astype(float)
)
df1.info()
df1.sample(5)
```

## Dataset 2 (prices in MXN → USD)

```python
df2 = pd.read_csv(path2)
df2.info()
df2.dropna(inplace=True)

df2.rename(columns={"price_mxn": "price_usd"}, inplace=True)
df2["price_usd"] = df2["price_usd"] / MXN_PER_USD
df2.info()
df2.head()
```

## Dataset 3 (split lat-lon, extract state)

```python
df3 = pd.read_csv(path3)
df3.info()
df3.dropna(inplace=True)

# Split "lat,lon" into two columns
df3[["lat", "lon"]] = df3["lat-lon"].str.split(",", expand=True)
df3["lat"] = df3["lat"].astype(float)
df3["lon"] = df3["lon"].astype(float)

# State is the 3rd item in "|México|Estado de México|Toluca|..."
df3["state"] = df3["place_with_parent_names"].str.split("|").str[2]

df3.drop(columns=["lat-lon", "place_with_parent_names"], inplace=True)
df3.info()
df3.head()
```

## Combine

```python
finalDf = pd.concat([df1, df2, df3], ignore_index=True)

# Optional: remove exact duplicate rows (6 found)
# finalDf.drop_duplicates(inplace=True)

finalDf.info()
finalDf
```

## Export (optional)
